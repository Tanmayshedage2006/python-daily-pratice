# Exception Handling

```python
try:
    a=int(input("Enter a "))
    b=int(input("Enter b"))
    print(a/b)
    arr=[10,20,30,40,50,]
    print(arr[8]) 
except ValueError:
    print("Enter Valid Value")
except IndexError:
    print("Index not avialble ")
except ZeroDivisionError:
    print("We can not divide by zero") 
```

```python
 try:
    a=int(input("Enter a "))
    b=int(input("Enter b"))
    print(a/b)
finally:
    print("I am in finally")
```

```python
try:
    a=int(input("Enter a "))
    b=int(input("Enter b"))
    print(a/b)
except:
    print("Exception")
else:
    print("I am in else")  # if the try block excute then else will run # if the try not excute then else will not run 
finally:
    print("I am in finally block")
```

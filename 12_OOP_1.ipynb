```python
class bank():
    def __init__(self,balance):
        self.balance = balance
    def intrest(self):
        return self.balance*0.10
    def fd(self,months):
        total =self.balance*(0.10*months)
        return total
```

# Incapsulation

```python
class car():
    def __init__(self,engine,fuel,seat):
        self.engine=engine # public
        self._fuel=fuel    # protect
        self.__seat=seat   # private
```

```python
c1 =car(2000,'Petrol',5)
```

```python
c1.engine # public
```

```python
c1._fuel #private
```

```python
c1._car__seat # private
```

# Inheritance

```python
class parent():
    def singing(self):
        return " I am parent class and having skills of singing"
```

```python
class child(parent):
    def cricket(self):
        return "I am child class and having skills of cricket"
```

```python
obj2 =child()
```

```python
obj2.cricket()
```

```python
obj2.singing()
```

```python
obj3 = parent()
```

```python
obj3.singing()
```

```python
class car():
    def __init__(self,engine,fuel,seat,color): 
        self.engine=engine
        self.fuel=fuel
        self.seat=seat
        self.color=color
    def start(self):
         return 'Car is Starting'
```

```python
class bus(car):
    def off(self):
        return" Bus is getting off"
```

```python
obj4 = bus(500,'Diesel',30,'Red')
```

```python
obj4.start()
```

```python
obj4.color
```

```python
obj4.off()
```

```python
class bike(bus):
    def kickstart(self):
        return 'Bike will start'
```

```python
# car ---> bus ---> bike
```

```python
obj6= bike(500,"disel",3,'red')
```

```python
obj6.kickstart()
```

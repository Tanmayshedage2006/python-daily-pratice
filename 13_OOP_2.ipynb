```python
from abc import ABC, abstractmethod
```

```python
class car(ABC):
    @abstractmethod
    def airbag(self):
        pass
```

```python
class nexon(car):
    def airbag(self):
        return " Airbags are installed in car"
    def button_start(Self):
        return 'Car will start'
```

```python
obj = nexon()
```

```python
obj.airbag()
```

```python
obj.button_start()
```

```python
class phone(ABC):
    @abstractmethod
    def call(self):
        pass
```

```python
class samsung(phone):
    def call(Self):
        return "calllin from samsung phone"
    def camera(self):
        return "opening camera from samsung mobile"
```

```python
s11 =samsung()
```

```python
s11.call()
```

```python
s11.camera()
```

```python
import re
txt="my age is 4 and my salary is 10000000"
result=re.findall(r"\d+",txt) #  /d = contain only one by one /d = contian all at a time 
print(result)                 # r = if you have to use / you have to print the r
```

```python
txt="cat dgi cow"
result=re.findall(r"cow|dog",txt)
print(result)
```

```python
txt="my age is 4 and my salary is 10000000"
result=re.findall(r"\D{9}",txt) # /D = is return all the strings 
print(result)
```

# lambda Function

```python
x=lambda a:a+10
print(x(5))
```

```python
x=lambda a,b:a*b
print(x(5,6))
```

```python
x=lambda a,b,c:a+b+c
print(x(5,6,2))
```

```python
e=lambda x,y,z:x/y/z
print(e(9,9,4))
```

```python
n=lambda x,y,z:x*y*z
print(e(7,8,6))
```

```python
def myfunc(n):
    return lambda a:a*n
mydoubler = myfunc(2)
print(mydoubler(11))
```

```python
def square(x):
    return x*x
def apply(func,value):
    return func(value)
print(apply(square,4))
```

```python
def double(x):
    return x*2
def calculate(func,n):
    return func(n) + 10
print(calculate(double,5))
```

# map

```python
numbers = [5,6,7,8,9]
def square(x):
    return x*x
result = map(square,numbers)
print(list(result))
```

```python
numbers=[1,2,3,4,5]
result=map(lambda x:x*x,numbers) # map means in simple what to do and where to do  # it tells before , and after , 
print(list(result))
```

# filter

```python
numbers=[1,2,3,4,5,6,7,8,]
result= filter(lambda x:x %2 == 0,numbers)
print(list(result))
```

```python
marks=[25,80,35,90,45,60]
passed=filter(lambda m:m>=40,marks)
print(list(passed))
```

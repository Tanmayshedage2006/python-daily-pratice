```python
def large_Sequence(n):
    for i in range(n):
        yield i
```

```python
def echo_genration():
    while True:
        recived = yield 
        print("Received",recived)
```

```python
def my_gen():
    try:
        yield 1
        yield 2
        yield 3
    finally:
        print("Genrator closed")
```

```python
gen = my_gen()
print(next(gen))
gen.close()
print(next(gen)) # because the excuetion is stop 
```

```python
import pandas as pd
```

```python
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Customer": [" rahul ", "AMAN", "Priya", "shad", "NEHA", "Aman"],
    "Price": ["2500", "Invalid", "3500", "4500", "2000", "3000"],
    "Order_Date": [
        "2024-01-10",
        "15/02/2024",
        "2024-03-20",
        "2024-04-15",
        "2024-05-10",
        "2024-06-01"
    ],
    "City": ["jaipur", "DELHI", "Pune", "JAIPUR", "mumbai", "delhi"],
    "Email": [
        "rahul@gmail.com",
        "aman@yahoo.com",
        "priya@gmail.com",
        "shad@outlook.com",
        "neha@gmail.com",
        "aman@yahoo.com"
    ]
}

df = pd.DataFrame(data)
```

```python
df['Customer'] = df['Customer'].str.strip().str.title()  # strip = remove space 
df                                                       # title = standardize names
```

```python
df['Price'] = pd.to_numeric(df['Price'], errors='coerce') # errors='coerce' if there is no value it wil be show NaN
d
```

```python
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
df
```

```python
df['City'] = df['City'].str.strip().str.title()
df
```

```python
print(df[df.duplicated('Email', keep=False)])
```

```python
df = df.drop_duplicates(subset='Email', keep='first')
df
```

```python
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df['Quarter'] = df['Order_Date'].dt.quarter

print(df[['Order_Date', 'Year', 'Month', 'Quarter']])
```

```python
df['Domain'] = df['Email'].str.split('@').str[1]

print(df[['Customer', 'Email', 'Domain']])
```

```python
gmail = df[df['Email'].str.contains('@gmail.com')]

print(gmail[['Customer', 'Email']])
```

```python
avg_price = df['Price'].mean()

print(avg_price)
```

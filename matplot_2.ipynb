```python
import pandas as pd

df = pd.read_csv("industry_retail_orders_dirty_1000.csv")
```

```python
daily_sales=df.groupby('order_date')['net_amount'].sum()
daily_sales
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
```

```python
df.dtypes
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,5))
plt.plot(daily_sales.index,
        daily_sales.values,label="Daily Revenue",color='red')
plt.legend()
plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

# python Decorators

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner
```

```python
@changecase
def myfunction():
    return "Hello Tanmay"
```

```python
@changecase 
def otherfunction():
    return "I am Gaurav"
    
```

```python
print(myfunction())
print(otherfunction())
```

```python
def my_decorator(func):
    def wrapper():
        print("hello")
        func()
    return wrapper
    
```

```python
def my_decorator(func):
    def wrapper():
        print("Start")
        func()
        print("End")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("HEllo ")
```

```python
say_hello()
```

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("----- Function Started -----")

        result = func(*args, **kwargs)

        print("----- Function Completed -----")
        return result

    return wrapper
```

```python

@my_decorator
def calculate_total(price, quantity, discount=0):
    total = price * quantity
    final_price = total - discount

    print("Price:", price)
    print("Quantity:", quantity)
    print("Discount:", discount)

    return final_price
    
answer = calculate_total(500, 3, discount=100)
print("Final Amount:", answer)
```

```python
def my_decorator(func):
    def wrapper(username, is_logged_in):
        print("Checking user authentication...")

        if is_logged_in:
            print("Welcome,", username)
            result = func(username, is_logged_in)
            print("Access completed successfully!")
            return result
        else:
            print("Access Denied! Please login first.")

    return wrapper
```

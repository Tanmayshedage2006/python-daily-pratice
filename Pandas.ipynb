```python
import pandas as pd  # why to use as 
```

```python
marks=pd.Series([10,30,40,30]) # one dimensional 
print(type(marks))
print(marks)
```

```python
data={
    "name":["Tanmay","Gaurav","Sarthak"], # two dimensional  data frame 
    "age":[23,23,23],
    "marks":[90,40,40]
}
df=pd.DataFrame(data)
print(df)
```

```python
data=[
     ["Tanmay",34,343433],
    ["Gaurav",34,342222],
    ["sahil",34,340300]
]                              # using list
df=pd.DataFrame(data,columns=["Name","Age","Salary"])
print(df)
```

```python
import numpy as np
```

```python
data=np.array([
     ["Tanmay",34,343433],
    ["Gaurav",34,342222],
    ["sahil",34,340300]            # using array 
] ) 
df=pd.DataFrame(data,columns=["Name","Age","Salary"])
print(df)
```

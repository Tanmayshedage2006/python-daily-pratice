```python
def even_odd(n):
    try:
        if n%2==0:
            return "even no",n
        else:
            return "odd ",n
    except Exception as e:
        print("Error in even_odd func")
        print(e)
```

```python
even_odd('h')
```

```python
def div(a,b):
    logging.info(f'value fo a{a} and b {b}')
    try:
        result=a/b
        logging.info(f'result for {a} and {b} is {result}')
        logging.info('Project complted')
        return result
    except Exception as e:
        logging.error("Error in div function")
        logging.error(e)
```
